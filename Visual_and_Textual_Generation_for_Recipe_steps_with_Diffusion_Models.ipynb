```python

# Visual and Textual Generation with Diffusion Models

# ## Objective
# This notebook demonstrates how to use Stable Diffusion and DALL-E models for generating creative visual and textual content.

# ### Steps
# 1. Initialize Stable Diffusion for image generation
# 2. Generate sample images based on text prompts
# 3. Implement DALL-E for generating descriptive text

# Importing necessary libraries
import torch
from diffusers import StableDiffusionPipeline
from transformers import pipeline
import matplotlib.pyplot as plt

# Initialize Stable Diffusion for image generation
pipe = StableDiffusionPipeline.from_pretrained('CompVis/stable-diffusion-v1-4')
pipe.to('cuda')

# Generate an image
prompt = "A person chopping carrots in the kitchen."
image = pipe(prompt).images[0]

plt.imshow(image)
plt.axis('off')
plt.show()

# DALL-E for text generation
text_generator = pipeline('text-generation', model='EleutherAI/gpt-neo-1.3B')
generated_text = text_generator("Describe a futuristic city", max_length=50, do_sample=True)[0]['generated_text']
print(generated_text)
```
